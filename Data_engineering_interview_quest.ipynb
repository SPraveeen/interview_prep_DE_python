## remove duplicates in a list

```python
nums=[1,2,3,2,4,1,5]
l=[]

for i in nums:
    if i not in l:
        l.append(i)

print(l)
```

## find largest element in alisr

```python
l=[2,3,44,1,55,6,7]
max=l[0]
for i in l:
    if i>max:
        max=i
print(max)
```

## count word freq in a string

```python
a="data engineering data python sql data engineering"
d=a.split(' ')
e={}
for i in d:
    if i in e:
        e[i]+=1
    else:
        e[i]=1
print(e)
```

## find factoriial of a number

```python
def factorial(n):
    if n==0:
        return 1
    else:
        return n*factorial(n-1)

print(factorial(5))
```

```python
def fact(n):
    result=1

    for i in range(1,n+1):
        result=result*i
    return result

print(fact(5))
```

## check if a no is prime

```python
def prime(n):
    if n<=1:
        return False
    for i in range(2,n):
        if n%i==0:
            return False
        return True
print(prime(12))
```
