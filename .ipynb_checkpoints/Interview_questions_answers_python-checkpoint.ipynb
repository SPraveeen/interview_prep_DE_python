## 1) write a python program to find out common letters between two strings 
## eg-> naina // reena

```python
a='naina'
b='reene'

def common_letters(a,b):
    c=set(a)
    d=set(b)
    return c&d

print(common_letters(a,b))
```

## 2) python program to count the frequency of words appearing in a string
## eg: sheena loves eating apple and mango. Her sister also loves eating apple and mango.

```python
a="sheena loves eating apple and mango. Her sister also loves eating apple and mango."
from collections import Counter
b=Counter(a)
b
```

```python
n={}
for i in a:
    if i in n:
        n[i]+=1
    else:
        n[i]=1
print(n)
```

```python
def freq_words(d):
    c=d.split(' ')
    d={}
    for i in c:
        if i not in d.keys():
            d[i]=0
        d[i] += 1
    return d
print(freq_words(a))
```

## 3) program to convert two list into a dictionary
### list1=['Naina','Kimi','Shemma']
### list2=[23456,67890,98754]

```python
list1=['Naina','Kimi','Shemma']
list2=[23456,67890,98754]
def list_to_dict(a,b):
    e=dict(zip(a,b))
    return e

print(list_to_dict(list1,list2))
    
```

```python
def dict_to_tuple():
    d={'Naina': 23456,'Kimi': 67890,'Shemma': 98754}
    for i in d.items():
        print(i)
        
dict_to_tuple()
```

## 4) Find missing number in an array(using summation and XOR operation)
## ex -> 1 2 4 5 6 7

```python
a=[1,2,4,5,6,7]
def find_missing_numbers(a):
    n=a[-1]
    sum1=0
    total=n*(n+1)//2
    sum1=sum(a)
    print(total-sum1)

find_missing_numbers(a)
```
